```python
#step 1 library import
import pandas as pd
import numpy as np
```

```python
##Load Uncleaned File
df=pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
##Check null value
df.isnull().sum()
```

```python
##Check duplicate value
df.duplicated().sum()
```

```python
##Remove duplicates value
df.drop_duplicates(inplace=True)
```

```python
##Replace value
df.replace(['N/A','Null',''],np.nan,inplace=True)
```

```python
df
```

```python
##Remove Extra Space
df=df.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

##Standardize specific columns

```python
df['Customer_Name']=df['Customer_Name'].str.title()
```

```python
df['City']=df['City'].str.title()
```

```python
df['State']=df['State'].str.title()
```

```python
df
```

```python
df=df[df['Email'].str.contains('@',na=False)].copy()
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
```

```python
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
cols=['Qty','Unit_Price','Discount']
for c in cols:
    df[c]=pd.to_numeric(df[c],errors='coerce')
```

```python
df=df[df['Qty']>0]
```

```python
##Filling Blank space
df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('Unknown')
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

###Features Engineering

```python
df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df['Profit']=df['Net_Amount']*0.20
```

```python
df["Year"]=df['Order_Date'].dt.year
```

```python
df['week_day']=df['Order_Date'].dt.day_name()
```

```python
df
```

##Loading cleaned data as CSV file

```python
df.to_csv('Clean_Ecommerce.csv',index=False)
```
